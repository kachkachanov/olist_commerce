```python
js = pd.merge(order_items, products, how='inner', on = 'product_id')
def outliers_clean(group):
    q1 = group.price.quantile(0.250) 
    q3 = group.price.quantile(0.750) 
    IQR = q3 - q1 
    return group[
        (group.price >= q1 - 1.5*IQR) &
        (group.price <= q3 + 1.5*IQR)
        ]
clean_price = js.groupby('category_name', group_keys=False).apply(outliers_clean)
print(len(js)-len(clean_price))
```

График с кривой Парето 

```python
import matplotlib.pyplot as plt

item_products = pd.merge(products, order_items, how='inner', on='product_id')
orders_products = pd.merge(orders, item_products, how='inner', on='order_id')
payments_products = pd.merge(payments, orders_products, how='inner', on='order_id')

group_products = payments_products.groupby('category_name')['payment_value'].sum().sort_values(ascending=False)

percent_group = group_products / group_products.sum()
pareto_products = percent_group.cumsum()

df = percent_group.reset_index(name='percent')
df['cum_percent'] = pareto_products.values

fig, ax1 = plt.subplots(figsize=(14, 6))

# столбики
ax1.bar(df['category_name'], df['percent'])
ax1.set_ylabel('Percent of revenue')
ax1.tick_params(axis='x', rotation=90)

# линия накопительного процента
ax2 = ax1.twinx()
ax2.axhline(0.8, linestyle='--')
ax2.plot(df['category_name'], df['cum_percent'], marker='o')
ax2.set_ylabel('Cumulative percent')

plt.title('Pareto chart by product category')
plt.show()
```

```python
def abc_class(cum_pct):
    if cum_pct <= 0.8:
        return 'A'
    elif cum_pct <= 0.95:
        return 'B'
    else:
        return 'C'

df['abc_class'] = df['cum_percent'].apply(abc_class)
print(df[['category_name', 'percent', 'cum_percent', 'abc_class']])

print(df['abc_class'].value_counts())
```

```python
monthly_demand = payments_products.groupby([
    payments_products['purchase_timestamp'].dt.to_period('M'),
    'category_name'
])['payment_value'].sum().reset_index()

xyz = monthly_demand.groupby('category_name')['payment_value'].agg(['mean', 'std']).reset_index()
xyz['cv'] = (xyz['std'] / xyz['mean'] * 100).round(2)

def xyz_class(cv):
    if cv <= 25:
        return 'X'  # стабильный спрос
    elif cv <= 50:
        return 'Y'  # умеренные колебания
    else:
        return 'Z'  # нестабильный спрос

xyz['xyz_class'] = xyz['cv'].apply(xyz_class)
print(xyz.sort_values('cv'))
```

```python
abc_xyz = pd.merge(df[['category_name', 'abc_class', 'percent']], 
                    xyz[['category_name', 'cv', 'xyz_class']], 
                    on='category_name')

abc_xyz['combined'] = abc_xyz['abc_class'] + abc_xyz['xyz_class']
print(abc_xyz.sort_values('percent', ascending=False))

# Сводная матрица
print(abc_xyz.groupby('combined').size().sort_values(ascending=False))
```

RFM на customer_unique_id

```python
ord_cst = pd.merge(orders_clean, customers, how = 'inner', on='customer_id') 
ord_cst_p = pd.merge(ord_cst, payments_clean, how = 'inner', on = 'order_id')
ord_cst_p 
```

```python
analysis_date = ord_cst_p['purchase_timestamp'].max() + pd.Timedelta(days=1)
```

```python
rfm = ord_cst_p.groupby('customer_unique_id').agg({
    'purchase_timestamp':'max',
    'order_id':'count',
    'payment_value':'sum'
})
```

```python
rfm.columns = ['LastPurchase', 'Frequency', 'Monetary']
```

```python
rfm['Recency'] = (
    analysis_date - rfm['LastPurchase']
).dt.days
```

```python
rfm = rfm[['Recency', 'Frequency', 'Monetary']]
rfm 
```

```python
rfm.groupby('Frequency').agg(
    customers=('Frequency', 'count'),
    avg_monetary=('Monetary', 'mean'),
    median_monetary=('Monetary', 'median')
)
```

```python
rfm['R'] = pd.qcut(rfm['Recency'], 5, labels=[5,4,3,2,1])

rfm['M'] = pd.qcut(rfm['Monetary'], 5, labels=[1,2,3,4,5])

rfm['F'] = np.where(rfm['Frequency']==1,1,2)
```

```python
def segment(row):

    if row.R >= 4 and row.F == 2 and row.M >= 4:
        return "Champions"

    elif row.R >= 4 and row.F == 2:
        return "Low-Value Repeat"

    elif row.R >= 4 and row.F == 1 and row.M >= 4:
        return "Potential Loyalists"

    elif row.R >= 4 and row.F == 1:
        return "New Customers"

    elif row.R <= 2 and row.M >= 4:
        return "Big Spenders at Risk"

    elif row.R <= 2:
        return "Lost"

    else:
        return "Regular"
```

```python
rfm['Segment'] = rfm.apply(segment, axis=1)
```

```python
rfm.groupby('Segment').agg(
    customers=('Recency','count'),
    revenue=('Monetary','sum'),
    avg_monetary=('Monetary','mean'),
    avg_recency=('Recency','mean'),
    repeat_rate=('Frequency', lambda x: (x > 1).mean())
)
```

A/B - testing

```python
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

orders_customers = pd.merge(orders_clean, customers, on='customer_id')
baseline_rpr = (orders_customers.groupby('customer_unique_id')['order_id'].nunique() > 1).mean()
print(f"Baseline RPR: {baseline_rpr:.4f}")

# MDE — минимальный эффект который хотим поймать (например +15% relative)
mde_relative = 0.15
target_rpr = baseline_rpr * (1 + mde_relative)

effect_size = proportion_effectsize(baseline_rpr, target_rpr)

power_analysis = NormalIndPower()
sample_size = power_analysis.solve_power(
    effect_size=effect_size, alpha=0.05, power=0.8, alternative='two-sided'
)
print(f"Sample size per group: {int(sample_size)}")

# Стратифицированное сплитование
import numpy as np
np.random.seed(42)

customers_unique = orders_customers[['customer_unique_id', 'state']].drop_duplicates()
customers_unique['group'] = customers_unique.groupby('state')['customer_unique_id'].transform(
    lambda x: np.random.choice(['control', 'test'], size=len(x))
)
print(customers_unique['group'].value_counts())

# SRM-check
from scipy.stats import chisquare
observed = customers_unique['group'].value_counts().values
chisquare(observed)
```
